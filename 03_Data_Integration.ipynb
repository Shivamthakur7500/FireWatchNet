```python
import pandas as pd
from pathlib import Path
```

#*Loading both the processed datasets*
* **Weather Dataset**
* **VIIRS Dataset**

#*Setting Project Directory*

```python
PROJECT_DIR = Path("/content/drive/MyDrive/FireWatchNet")
```

#**Loading both processed datasets**

```python
viirs_path = (
    PROJECT_DIR
    / "Data/processed/VIIRS/VIIRS_daily_fire_labels_2023_2024.csv"
)

weather_path = (
    PROJECT_DIR
    / "Data/processed/Weather/Weather_2023_2024_cleaned.csv"
)

viirs = pd.read_csv(viirs_path)
weather = pd.read_csv(weather_path)

print("VIIRS shape:", viirs.shape)
print("Weather shape:", weather.shape)

display(viirs.head())
display(weather.head())
```

#*Converting date columns from both the Datasets*

```python
viirs["date"] = pd.to_datetime(
    viirs["date"],
    errors="coerce"
)

weather["date"] = pd.to_datetime(
    weather["date"],
    errors="coerce"
)
```

#*Removing unnecessary weather columns*

```python
weather = weather.drop(
    columns=["system:index", ".geo"],
    errors="ignore"
)

display(weather.head())
print(weather.columns.tolist())
```

#*Merge both the Datasets*

```python
integrated_data = pd.merge(
    weather,
    viirs,
    on="date",
    how="inner"
)

print("Integrated shape:", integrated_data.shape)
display(integrated_data.head())
```

```python
print(integrated_data.isna().sum())
print("Duplicate dates:", integrated_data["date"].duplicated().sum())
print(integrated_data.shape)
```

#*Saving the Merged Datasets*

```python
integration_dir = (
    PROJECT_DIR / "Data/processed/Integrated"
)

integration_dir.mkdir(
    parents=True,
    exist_ok=True
)

integrated_path = (
    integration_dir
    / "FireWatchNet_Integrated_2023_2024.csv"
)

integrated_data.to_csv(
    integrated_path,
    index=False
)

print("Saved:", integrated_path)
```
