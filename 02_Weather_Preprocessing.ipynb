```python
weather_path = (
    PROJECT_DIR
    / "/content/drive/MyDrive/FireWatchNet/Data/raw/fire/Weather/Uttarakhand_Weather_2023_2024.csv"
)

weather = pd.read_csv(weather_path)

print("Shape:", weather.shape)
display(weather.head())
print(weather.columns.tolist())
```

```python
print(weather.info())
print(weather.isna().sum())
```

```python
weather.columns = (
    weather.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(weather.columns.tolist())
```

```python
weather["date"] = pd.to_datetime(
    weather["date"],
    errors="coerce"
)

print("Minimum date:", weather["date"].min())
print("Maximum date:", weather["date"].max())
```

```python
print("Duplicate dates:", weather["date"].duplicated().sum())
```

```python
weather_output_dir = (
    PROJECT_DIR / "Data/processed/Weather"
)

weather_output_dir.mkdir(
    parents=True,
    exist_ok=True
)

weather_output_path = (
    weather_output_dir
    / "Weather_2023_2024_cleaned.csv"
)

weather.to_csv(
    weather_output_path,
    index=False
)

print("Saved:", weather_output_path)
```
