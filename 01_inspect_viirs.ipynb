#*01. Import libraries*

```python
import pandas as pd
from pathlib import Path

viirs_output_dir = Path("../Data/processed/VIIRS")
viirs_output_dir.mkdir(parents=True, exist_ok=True)
```

#*02. Load VIIRS data**


```python
file_path = Path(
    "/content/drive/MyDrive/FireWatchNet/Data/raw/fire/VIIRS/2023-2024/fire_archive_J1V-C2_798353.csv"
)

df = pd.read_csv(file_path)

print(df.shape)
display(df.head())
```

#*03. Inspect columns and rows*

```python
print(df.columns.tolist())
print(df.info())
print(df.isna().sum())
```

#*04. Clean column names*

```python
df.columns = df.columns.str.strip().str.lower()

print(df.columns.tolist())
```

#*05. Convert dates and coordinates*

```python
df["acq_date"] = pd.to_datetime(
    df["acq_date"],
    errors="coerce"
)

df["latitude"] = pd.to_numeric(
    df["latitude"],
    errors="coerce"
)

df["longitude"] = pd.to_numeric(
    df["longitude"],
    errors="coerce"
)
```

#*06. Remove missing/duplicate records*





```python
df = df.dropna(
    subset=["acq_date", "latitude", "longitude"]
)

df = df.drop_duplicates()

print("Remaining rows:", len(df))
```

#*07. Filter Uttarakhand*

```python
df = df[
    (df["acq_date"] >= "2023-01-01") &
    (df["acq_date"] < "2025-01-01")
]

print(df["acq_date"].min())
print(df["acq_date"].max())
```

```python
import pandas as pd
from pathlib import Path

# Load the filtered VIIRS dataset
input_file = Path(
    "/content/drive/MyDrive/FireWatchNet/Data/processed/VIIRS/"
    "VIIRS_Uttarakhand_2023_2024.csv"
)

output_dir = input_file.parent
output_file = output_dir / "VIIRS_daily_fire_labels_2023_2024.csv"

viirs = pd.read_csv(input_file)
viirs["acq_date"] = pd.to_datetime(viirs["acq_date"])

# Count satellite fire detections per day
daily_counts = (
    viirs.groupby("acq_date")
    .size()
    .rename("fire_count")
)

# Include every day, including days with zero detections
all_dates = pd.date_range("2023-01-01", "2024-12-31", freq="D")

daily = daily_counts.reindex(all_dates, fill_value=0).rename_axis("date").reset_index()

# Binary label: 1 = at least one detection; 0 = none
daily["fire_label"] = (daily["fire_count"] > 0).astype(int)

# Save as a new file
daily.to_csv(output_file, index=False)

print("Total days:", len(daily))
print("\nLabels:")
print(daily["fire_label"].value_counts().sort_index())
print("\nYearly summary:")
print(daily.groupby(daily["date"].dt.year).agg(
    total_days=("date", "size"),
    total_detections=("fire_count", "sum"),
    days_with_detections=("fire_label", "sum")
))
print("\nSaved to:", output_file)
```

```python
import pandas as pd
from pathlib import Path

BASE_DIR = Path("/content/drive/MyDrive/FireWatchNet/Data")

weather_file = (
    BASE_DIR / "processed/Weather/Weather_2023_2024_cleaned.csv"
)

labels_file = (
    BASE_DIR / "processed/VIIRS/VIIRS_daily_fire_labels_2023_2024.csv"
)

output_file = (
    BASE_DIR / "processed/Integrated/FireWatchNet_Integrated_2023_2024.csv"
)

# Load datasets
weather = pd.read_csv(weather_file)
labels = pd.read_csv(labels_file)

weather["date"] = pd.to_datetime(weather["date"])
labels["date"] = pd.to_datetime(labels["date"])

# Keep only the columns needed from the weather dataset
weather_columns = [
    "date",
    "temperature_2m",
    "total_precipitation_sum",
    "volumetric_soil_water_layer_1",
    "wind_speed"
]

weather = weather[weather_columns]

# Merge by date
integrated = weather.merge(
    labels[["date", "fire_count", "fire_label"]],
    on="date",
    how="outer",
    validate="one_to_one",
    indicator=True
)

# Validate the merge before saving
print("Merge status:")
print(integrated["_merge"].value_counts())

print("\nDataset shape:", integrated.shape)
print("Date range:", integrated["date"].min(), "to", integrated["date"].max())
print("\nMissing values:")
print(integrated.isna().sum())

# Save only if every date matched and no required values are missing
if (
    len(integrated) == 731
    and (integrated["_merge"] == "both").all()
    and integrated[weather_columns[1:] + ["fire_count", "fire_label"]]
        .notna().all().all()
):
    integrated = integrated.drop(columns="_merge")
    output_file.parent.mkdir(parents=True, exist_ok=True)
    integrated.to_csv(output_file, index=False)
    print("\nSUCCESS: Integrated dataset saved to:")
    print(output_file)
else:
    print("\nNOT SAVED: Investigate the merge or missing values first.")
```
